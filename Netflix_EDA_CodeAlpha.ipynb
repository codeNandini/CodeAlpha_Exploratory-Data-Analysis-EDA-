```python
movies = df_clean[df_clean['type'] == 'Movie'].copy()

movies['duration_num'] = pd.to_numeric(
    movies['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

print("Movie Duration Statistics:")
print(movies['duration_num'].describe())

plt.figure(figsize=(10, 5))
sns.histplot(movies['duration_num'].dropna(), bins=30, kde=True)

plt.title('Distribution of Movie Durations')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.show()

print("Shortest Movie:", movies['duration_num'].min(), "minutes")
print("Longest Movie:", movies['duration_num'].max(), "minutes")
print("Average Duration:", round(movies['duration_num'].mean(), 2), "minutes")
```

```python
from scipy.stats import binomtest

movie_count = (df_clean['type'] == 'Movie').sum()
total_count = df_clean['type'].count()

result = binomtest(
    movie_count,
    total_count,
    p=0.5,
    alternative='greater'
)

print("Movie Count:", movie_count)
print("Total Titles:", total_count)
print("Movie Proportion:", round(movie_count / total_count, 4))
print("P-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject H0: Movies make up significantly more than 50% of titles.")
else:
    print("Fail to reject H0: Insufficient evidence that Movies exceed 50%.")
```

```python
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(
    df_clean['type'],
    df_clean['rating']
)

print("Contingency Table:")
print(contingency_table)

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Content type and rating are significantly associated.")
else:
    print("Fail to reject H0: Insufficient evidence of an association.")
```

```python
Q1 = movies['duration_num'].quantile(0.25)
Q3 = movies['duration_num'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = movies[
    (movies['duration_num'] < lower_bound) |
    (movies['duration_num'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))

plt.figure(figsize=(10, 5))
sns.boxplot(x=movies['duration_num'])

plt.title('Boxplot of Movie Durations')
plt.xlabel('Duration (Minutes)')
plt.show()
```
